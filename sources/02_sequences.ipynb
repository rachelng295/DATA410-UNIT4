```python
b_set = {2, 5, 6, 8, 15, 32, 4, 6, 2, 15}
b_set
```

```python
c_tuple = (2, 5, 6, 8, 15, 32, 4, 6, 2, 15) #you can not change the values in a tuple, it is immutable
c_tuple
```

```python
d_fr_set = frozenset(b_set) # frozenset is an immutable version of a set, meaning that once it is created, its elements cannot be changed or modified. This can be useful in situations where you want to ensure that the set remains constant and cannot be altered accidentally.
d_fr_set
```
