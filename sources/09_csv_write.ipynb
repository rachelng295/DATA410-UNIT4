```python
import csv
```

```python
makes = ["Toyota", "Ford", "BMW"]
models = ["RAV4", "F150", "X6"]
max_speeds = [220, 200, 260]
powers = [195, 400, 368]
```

```python
car_list = list(zip(makes, models, max_speeds, powers))
car_list
```

```python
with open('../datasets/cars.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(["Make", "Model", "Max Speed", "Power"])
    writer.writerows(car_list)
```
