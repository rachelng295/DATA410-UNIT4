```python
import statistics as st
```

```python
values = [92, 64, 87, 89, 51, 76, 100, 70]
```

```python
print(st.mean(values)) #statistics.mean is a built-in function that calculates the mean of a list of numbers
```

```python
print(st.stdev(values)) #statistics.stdev is a built-in function that calculates the standard deviation of a list of numbers
```

```python
print(st.median(values)) #statistics.median is a built-in function that calculates the median of a list of numbers
```
